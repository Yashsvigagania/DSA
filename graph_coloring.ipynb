```python
n,m=map(int,input().split())
g=[list(map(int,input().split())) for _ in range(n)]
c=[0]*n
def f(i):
  if i==n:
    print(*c,end=" ")
    return 1
  for x in range(1,m+1):
    if all(not g[i][j] or c[j]!=x for j in range(n)):
      c[i]=x
      if f(i+1): return 1
  return 0
_ = f(0)
```
