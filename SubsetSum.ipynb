```python
n,s=map(int,input().split())
a=list(map(int,input().split()))
f=0
def backtrack(i,cur,res):
  global f
  if cur==s:
    print("["+" ".join(map(str,res))+ "]")
    f=1
    return
  if i==n or cur>s: return
  backtrack(i+1,cur+a[i],res+[a[i]])
  backtrack(i+1,cur,res)
backtrack(0,0,[])
if not f: print("No subset found")
```
